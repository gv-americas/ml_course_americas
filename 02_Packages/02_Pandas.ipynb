# The pandas Series

Pandas series will share many methods and atributes with numpy arrays, because they are numpy arrays under the hood. But there are some differences.

Key Differences:

* Indexing: Series support both integer-based and label-based indexing, while NumPy arrays primarily use integer-based indexing.
* Missing values: Series can handle missing values (NaN) gracefully, while NumPy arrays often require explicit handling of missing data.
* Data alignment: Series automatically align data based on their indices during operations, while NumPy arrays require manual alignment.

```python
s = pd.Series(['LIM', 'LIM', 'SAP', 'BRK', 'BRK'])
```

# Iterating over a DataFrame

```python
for i in df:
    print(i)
```

```python
for i in range(df.shape[0]):
    print(df.iloc[i])
```

* iterrows(): Returns an iterator yielding (index, Series) pairs for each row.
* itertuples(): Returns an iterator yielding namedtuples for each row, providing access to values by column name or position.

```python
for idx, row in df.iterrows():
    print(idx)
    print(row)
    print(row['col1'])
```

```python
for row in df.itertuples():
    print(row.col1)
```
