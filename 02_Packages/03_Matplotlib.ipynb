# Configuring the figure

https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set.html

```python
fig, ax = plt.subplots(figsize=(5,5))

ax.plot(x, y)

ax.set_title('Title'),
ax.set_xlabel('X Axis'),
ax.set_ylabel('Y axis'),
ax.set_xticks(np.arange(0, 10.1, 0.8)),
ax.set_ylim(0., 8),
ax.set_xlim(0, 10),
ax.set_aspect('equal'),


plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5,5))

ax.plot(x, y)

fig_settings = {
'title':'Title',
'xlabel':'X Axis',
'ylabel':'Y Axis',
'xticks':np.arange(0, 10.1, 0.8),
'ylim':(0., 8),
'xlim':(0, 10),
'aspect':'equal',
}

ax.set(
**fig_settings #pwe use double asterix to pass a dictionary as the arguments
)

plt.show()
```

## Functions

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html

```python
def exponetial(h, r, c0, b=0):
    a = r/3
    return b + c0 * (1 - np.exp(-h/a))
```

```python
x = np.arange(0, 100, 1)
y = exponetial(x, 60, 1)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x, y)

ax.set(
title='Exponential Variogram',
xlabel='h (m)',
ylabel='Variance'
)

ax.grid(linestyle='--')

plt.show()
```

### Location maps

```python
x = walker['X']
y = walker['Y']
v = walker['V']
```

```python
fig, ax = plt.subplots(figsize=(10, 10))

scatvu = ax.scatter(x, y, c=v, cmap='jet')

ax.set(
title = 'Location map for V',
xlabel = 'X (m)',
ylabel='Y (m)',
aspect = 'equal' # important to not get your maps distorted
)

ax.minorticks_on()
ax.grid(linestyle='--', which='major')

fig.colorbar(scatvu, label='Teor V')

plt.show()
```

```python
v = walker['T']
```

```python
fig, ax = plt.subplots(figsize=(10, 10))

ncats = np.unique(v).size
cat_labels = np.unique(v)

#transforming alphanum categories names to numerical only
catdic = {j:i for i,j in enumerate(cat_labels)} 
catnum = v.map(catdic) 
vlim = (0, ncats)

catcmap = plt.cm.get_cmap('jet', ncats) #creating our own colorscale

scatvu = ax.scatter(x, y, c=catnum, cmap=catcmap, vmin=vlim[0], vmax=vlim[1])

ax.set(
title = 'Locatoin map for T',
xlabel = 'X (m)',
ylabel='Y (m)',
aspect = 'equal'
)

ax.minorticks_on()
ax.grid(linestyle='--', which='major')

ticks_location = np.arange(ncats) + 0.5

cbar = fig.colorbar(scatvu, label='Cat T', ticks=ticks_location)
cbar.set_ticklabels(cat_labels)

plt.show()
```

```python
def locmap(x, y, v, cat, figsize, title, outfl):
    
    #creating the figure
    fig, ax = plt.subplots(figsize=figsize)

    #configuring the figure
    ax.set(
    title = title,
    xlabel = 'X (m)',
    ylabel='Y (m)',
    aspect = 'equal'
    )

    #adding elements
    ax.minorticks_on()
    ax.grid(linestyle='--', which='major')

    #checking if the variable to plot is categorical or numerical
    if cat:
        ncats = np.unique(v).size
        cat_labels = np.unique(v)

        #transforming alphanum categories names to numerical only
        catdic = {j:i for i,j in enumerate(cat_labels)} 
        catnum = v.map(catdic) 
        vlim = (0, ncats)

        catcmap = plt.cm.get_cmap('jet', ncats) #creating our own colorscale

        #ploting
        scatvu = ax.scatter(x, y, c=catnum, cmap=catcmap, vmin=vlim[0], vmax=vlim[1])
        
        #defining colorscale
        ticks_location = np.arange(ncats) + 0.5

        cbar = fig.colorbar(scatvu, label='Cat', ticks=ticks_location)
        cbar.set_ticklabels(cat_labels)


    else:
        #ploting
        scatvu = ax.scatter(x, y, c=v, cmap='jet')

        #defining colorscale
        fig.colorbar(scatvu, label='Grades')

    #saving
    plt.savefig(outfl, bbox_inches='tight', facecolor='white') 
```

```python
locmap(x=walker['X'], y=walker['Y'], v=walker['U'], cat=False, figsize=(10,10), title='Mapa U', outfl='mapa_u.png')
```

```python
locmap(x=walker['X'], y=walker['Y'], v=walker['T'], cat=True, figsize=(10,10), title='Mapa T', outfl='mapa_t.png')
```

## Grids

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imshow.html

```python
v = np.random.random((100,100))
```

```python
fig, ax = plt.subplots(figsize=(10, 10))

scatvu = ax.imshow(v, cmap='jet', extent=(0-1/2, 100+1/2, 0-1/2, 100+1/2), origin='lower')

ax.set(
title = 'Grid random noise',
xlabel = 'X (m)',
ylabel='Y (m)',
aspect = 'equal' #importante to grids
)

ax.minorticks_on()
ax.grid(linestyle='--', which='major')

# Create an axs for colorbar. The position of the axs is calculated based on the position of ax.
# You can change 0.01 to adjust the distance between the main image and the colorbar.
# You can change 0.02 to adjust the width of the colorbar.
# This practice is universal for both subplots and Geoaxs.
# https://stackoverflow.com/questions/18195758/set-matplotlib-colorbar-size-to-match-graph

cax = fig.add_axes([ax.get_position().x1+0.01,ax.get_position().y0,0.02,ax.get_position().height])
plt.colorbar(scatvu, cax=cax, label='Grades %') # Similar to fig.colorbar(im, cax = cax)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(15, 15))

scatvu = ax.imshow(v, cmap='jet', extent=(0-2*1/2, 200+2*1/2, 0-1/2, 100+1/2), origin='lower') #if we have different block dimensions in x and y

ax.set(
title = 'Grid random noise',
xlabel = 'X (m)',
ylabel='Y (m)',
aspect = 'equal' #importante para grids
)

ax.minorticks_on()
ax.grid(linestyle='--', which='major')

# Create an axs for colorbar. The position of the axs is calculated based on the position of ax.
# You can change 0.01 to adjust the distance between the main image and the colorbar.
# You can change 0.02 to adjust the width of the colorbar.
# This practice is universal for both subplots and Geoaxs.
# https://stackoverflow.com/questions/18195758/set-matplotlib-colorbar-size-to-match-graph

cax = fig.add_axes([ax.get_position().x1+0.01,ax.get_position().y0,0.02,ax.get_position().height])
plt.colorbar(scatvu, cax=cax, label='Grades %') # Similar to fig.colorbar(im, cax = cax)

plt.show()
```

### Bar and pizza plots

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.bar.html

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.pie.html

```python
v = walker['T']
```

```python
fig, ax = plt.subplots(figsize=(12,8))

x_labels = np.unique(v)
x_ticks = np.arange(len(x_labels))

heights = [np.sum(v==i) for i in x_labels]

ax.bar(x_ticks, heights)

for x, y in zip(x_ticks, heights):
    ax.annotate(str(y), (x, y), ha='center')

ax.set(
title='Categories',
ylabel='Number of samples',
xticks=x_ticks,
)
ax.set_xticklabels(x_labels, rotation=90) #definindo xticklabes foutside set method to use rotation argument

ax.grid(linestyle='--', axis='y')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,8))

labels = np.unique(v)

sizes = [np.sum(v==i) for i in x_labels]

ax.pie(sizes, labels=labels)

plt.show()
```

### 3d Scatter

https://matplotlib.org/stable/gallery/mplot3d/subplot3d.html

```python
fe = pd.read_csv(r'..\00_Data\carajas.csv')
```

```python
fe.head()
```

```python
x, y, z = fe['X'], fe['Y'], fe['Z']
v = fe['Fe']
```

```python
fig = plt.figure(figsize=(10,10))

ax = fig.add_subplot(111, projection='3d') #adding 3d ax

z_exageration = 2

ax.set_box_aspect((np.ptp(x), np.ptp(y), z_exageration*np.ptp(z)))  # aspect ratio is 1:1:1 in data space

ts = ax.scatter(x, y, z, c=v, s=2, cmap='jet')

ax.set(
ylabel='Y (m)',
xlabel='X (m)',
zlabel='Z (m)',
)

plt.colorbar(ts, label='Fe grades (%)', fraction=0.03, pad=0.08)

plt.show()
```
