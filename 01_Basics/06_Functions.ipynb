# Functions

A function in Python is a self-contained block of code that you can use over and over again.

```python
def my_sum(a, b):
    a+b
```

```python
r = my_sum(2,2)
```

```python
r
```

```python
def my_sum(a, b):
    return a+b
```

```python
r = my_sum(2,2)
```

```python
r
```

```python
def my_sum(a, b=2):
    return a+b
```

```python
my_sum(2)
```

```python
func_args = {
    'a':2,
    'b':3
}
```

```python
my_sum(**func_args)
```

```python
def my_sum(*nums):
    total = 0 

    print(type(nums))
    
    for i in nums:
        total += i
    
    return total 
```

```python
my_sum(1,2,3,4,5)
```

```python
def my_fun():
    print('Hello World!')
```

```python
my_fun()
```
