```python
# It is also possible to convert a string to a datetime object with `strptime`
date_string = "02/01/2024"
date = datetime.strptime(date_string, "%d/%m/%Y") # This format must match with the string you entered
print(type(date), date)
```

```python
# It is also possible to convert a datetime object to a string with `strftime`
date = datetime.now()
date_string = date.strftime("%A %d %B %Y")
print(type(date_string), date_string)
```

You can find a list of the different datetime formats [here](https://strftime.org/).
