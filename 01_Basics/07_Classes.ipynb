## Atributes

Think of attributes as characteristics.

```python
class Parrot:
    def __init__(self, name, color):
        self.name = name
        self.color = color
```

```python
parrot1 = Parrot('Luna', 'Blue')
```

```python
parrot1.name
```

```python
parrot1.name = 'Suzy'
```

```python
parrot1.name
```

```python
parrot2 = Parrot('Toby', 'Green')
```

```python
parrot2.color
```

# Methods

Think of methods as actions (or functions of classes).

```python
class Parrot:
    def __init__(self, name, color):
        self.name = name
        self.color = color

    def chirp(self):
        print('TWEEEEET!!!!')
```

```python
parrot3 = Parrot('Rex', 'Green')
```

```python
parrot3.chirp()
```

```python
class Parrot:
    def __init__(self, name, color):
        self.name = name
        self.color = color

    def chirp(self):
        print('TWEEEEET!!!!')
    
    def repeat(self, word):
        print(f'TWEEET! {word}!')
```

```python
parrot4 = Parrot('Peanuts', 'Grey')
```

```python
parrot4.repeat('Hello world')
```

# Contextualization

```python
class Grid:
    def __init__(self, ox, oy, nx, ny, sx, sy):
        self.ox = ox
        self.oy = oy

        self.nx = nx
        self.ny = ny

        self.sx = sx
        self.sy = sy
```

```python
estimation_grid = Grid(0, 0, 10, 10, 5, 5)
```

```python
simulation_grid = Grid(0, 0, 50, 50, 1, 1)
```
