### Useful methods

* capitalize(): Converts the first character to upper case
* count(): Returns the number of times a specified value occurs in a string

* format(): Formats specified values in a string (it is better to use the f string, explained in the next section)
* isalnum(): Returns True if all characters in the string are alphanumeric
* isalpha(): Returns True if all characters in the string are in the alphabet
* islower(): Returns True if all characters in the string are lower case
* isnumeric(): Returns True if all characters in the string are numeric
* isupper(): Returns True if all characters in the string are upper case
* replace(): Returns a string where a specified value is replaced with a specified value
* split(): Splits the string at the specified separator, and returns a list
* upper(): Converts a string into upper case

### r strings

Means "raw". Useful for strings which have escape characters in them like "\\". Used notably for file paths.

```python
path = "C:\User\public\example_file_path.txt"
print(path)
```

```python
path = r"C:\User\public\example_file_path.txt"
print(path)
```

### Combining f and r strings

```python
file_name = "mining_area.shp"
path = fr'C:\user\public\{file_name}'
print(path)
```

* Non-empty objects: Most non-empty objects like strings, lists, dictionaries, and sets are considered True.
* Empty objects: Empty sequences like empty strings, lists, and sets are considered False.
* Numbers: Non-zero numbers are True, while zero is False.
* None: The special value None is always False.
* Booleans: True is True, and False is False.
