## Useful functions

* enumerate(): Iterates over a sequence (list, string, etc.) and returns an enumerated object containing two elements for each item: the index and the item itself
* range(): Generates a sequence of numbers within a specified range
* zip(): Combines elements from two or more iterables into tuples, pairing corresponding elements from each iterable

```python
for i, j in enumerate(['data', 'mine']):
    print(i, j)
```

```python
for i, j in zip(['data', 'mine'], ['software', '2026']):
    print(i, j)
```

# Flow modifiers

* break statement: Immediately exits a loop or nested block
* continue statement: Skips the current iteration of a loop and moves to the next one
* pass statement: Does nothing but acts as a placeholder when a statement is required syntactically but no action is needed

```python
for i in range(10):

    if i%2 == 0:
        continue
    elif i%3 == 0:
        pass
    elif i%5 == 0:
        break
    else:
        print(i)

    print(i)
```
